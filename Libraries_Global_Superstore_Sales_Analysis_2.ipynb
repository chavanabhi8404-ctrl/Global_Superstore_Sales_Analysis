## step 3: Exploratory Data Analysis (EDA)

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

# Q1. What is the distribution of Sales?

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Sales'], bins=50, kde=True)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

## Analysis:
- Sales distribution is right-skewed.
- Most transactions have relatively low sales values.

## Conclusion:
- Most sales transactions generate smaller revenue, while a few high-value orders contribute significantly to total sales.

# Q2. Which Category generates the highest Sales?

```python
category = df.groupby('Category')['Sales'].sum()

print(category)
category.plot(kind='bar', title='Category-wise Sales')
plt.ylabel('Sales')
plt.show()
```

# Analysis:
- Technology generates the highest sales.
- Furniture generates comparatively lower sales.

# Conclusion: 
- Technology is the top sales-generating category.

# Q3. Which Customer Segment generates the highest Sales?

```python
segment_sales = df.groupby('Segment')[['Sales','Profit']].sum()

print(segment_sales)
segment_sales['Sales'].plot(kind='pie',autopct='%1.1f%%',figsize=(7,7))
plt.title("Customer Segment Sales Distribution")
plt.ylabel("")
plt.show()
```

# Analysis:
- Consumer segment generates the highest sales.
- Corporate segment generates the second-highest sales.
- Home Office generates comparatively lower sales.

# Conclusion:

- Consumer customers are the major revenue contributors, so the business should focus on customer retention and engagement in this segment.

# Q4. Which Region has the highest Profit?

```python
region = df.groupby('Region')['Profit'].sum()

print(region)
region.plot(kind='bar', title='Region-wise Profit')
plt.ylabel('Profit')
plt.show()
```

# Analysis:
- West region generates the highest profit.
- Central region generates comparatively lower profit.

# Conclusion: 
- West is the most profitable region.

# Q5. What is the correlation among Sales, Profit, Quantity and Discount?

```python
corr = df[['Sales','Profit','Quantity','Discount']].corr()

plt.figure(figsize=(8,5))
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

# Analysis:
- Discount and Profit have a negative correlation (approximately -0.22).
- Quantity has a weak positive correlation with Profit.

# Conclusion:
- Sales have a positive relationship with Profit, while higher discounts tend to be associated with lower profitability.

# Q6. What is the relationship between Sales and Profit?

```python
sns.scatterplot(data=df, x='Sales', y='Profit')
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

# Analysis:
- The scatter plot shows the relationship between Sales and Profit.
- Most points show a positive relationship.

# Conclusion:
- Higher sales do not always guarantee higher profit. The company should focus on profitable sales.


# Q7. Which Category has the highest variation in Profit?

```python
sns.boxplot(data=df, x='Category', y='Profit')
plt.title('Category-wise Profit Distribution')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.show()
```

# Analysis:
- The box plot compares profit distribution across categories.
- It shows the median, spread, and potential outliers.

# Conclusion:
- The company should identify categories with frequent losses and improve their profitability.
